# Session 9 — Observable loss harness

This notebook executes the complete PyTorch harness from `loss_harness.py`. It prints tensor shapes with dimension meanings, verifies the next-token shift using token strings, demonstrates padding and document-boundary masks, checks untrained perplexity, compares tied/untied heads, measures ordinary versus chunked cross-entropy memory, and trains an additional t+2 head.

The corpus is embedded in the harness, so no dataset download is required.

```python
from pathlib import Path

# When run from the repository, use the adjacent source file. When opened
# directly in Colab, fetch the same reviewed source from this GitHub repo.
source_path = Path('loss_harness.py')
if source_path.exists():
    harness_source = source_path.read_text()
else:
    import urllib.request
    raw_url = 'https://raw.githubusercontent.com/gmrammohan15/ERA5/main/session9/loss_harness.py'
    harness_source = urllib.request.urlopen(raw_url).read().decode('utf-8')
    source_path.write_text(harness_source)

namespace = {'__name__': '__notebook__'}
exec(compile(harness_source, str(source_path), 'exec'), namespace)
results = namespace['run']()
```

## Reading the output

The intentionally low same-position diagnostic loss is the off-by-one trap. The correct objective predicts the token to the right. Padding and cross-document transitions are excluded by loss masks, not silently treated as ordinary training examples. Chunking should preserve the loss while reducing peak memory. The t+2 curve is reported separately because it is a farther, less locally determined prediction.
